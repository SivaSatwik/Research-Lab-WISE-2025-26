```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('../../data/cleaned/cleaned_data.csv')
print(f"Dataset: {len(df):,} students × {len(df.columns)} columns")
```

```python
# SUMMARY STATISTICS FOR KEY VARIABLES
vars_info = {
    'Trust (Q15)': 'Q15',
    'Recommendation (Q16)': 'Q16',
    'Usage Intensity (Q18)': 'usage_intensity',
    'Capabilities (Q22)': 'capabilities_score',
    'Ethical Concerns (Q28)': 'ethics_score',
    'Attitudes (Q21)': 'attitudes_score',
}
rows = []
for name, col in vars_info.items():
    s = df[col].dropna()
    rows.append({
        'Variable': name, 'N': len(s), 'Mean': s.mean(),
        'SD': s.std(), 'Min': s.min(), 'Median': s.median(),
        'Max': s.max(), 'Missing%': (df[col].isna().sum()/len(df))*100
    })
summary = pd.DataFrame(rows).set_index('Variable')
print("Summary Statistics:")
print(summary.round(2).to_string())
```

```python
# STUDY LEVEL DISTRIBUTION
print("Study Level Distribution (Q8):")
sl = df['study_level'].value_counts()
for level in ['Undergraduate', 'Postgraduate/Masters', 'Doctoral']:
    if level in sl.index:
        print(f"  {level:25s}: {sl[level]:>6,} ({sl[level]/len(df)*100:5.1f}%)")
```

```python
# USAGE INTENSITY GROUPS (Q18 tertiles)
print("Usage Intensity Groups (Q18 tertiles):")
ug = df['usage_group'].value_counts().sort_index()
for grp in ug.index:
    print(f"  {grp:15s}: {ug[grp]:>6,} ({ug[grp]/df['usage_group'].notna().sum()*100:5.1f}%)")
```

```python
# KEY VARIABLES BY USAGE GROUP
print("Key Variables by Usage Group:")
for name, col in vars_info.items():
    print(f"\n{name}:")
    g = df.groupby('usage_group')[col].agg(['count','mean','std']).round(2)
    print(g.to_string())
```

```python
# KEY VARIABLES BY STUDY LEVEL
print("Key Variables by Study Level:")
for name, col in vars_info.items():
    print(f"\n{name}:")
    g = df.groupby('study_level')[col].agg(['count','mean','std']).round(2)
    print(g.to_string())
```

```python
# DISCIPLINE DISTRIBUTION (Q10)
print("Discipline Distribution (Q10):")
dd = df['discipline'].value_counts()
for d in dd.index:
    print(f"  {d:30s}: {dd[d]:>6,} ({dd[d]/df['discipline'].notna().sum()*100:5.1f}%)")
```

```python
# REGION/LANGUAGE DISTRIBUTION
print("Region/Language Distribution:")
rd = df['region'].value_counts()
for r in rd.index:
    print(f"  {r:15s}: {rd[r]:>6,} ({rd[r]/len(df)*100:5.1f}%)")

print("\n✓ Descriptive analysis complete")
```

## Figures

```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import seaborn as sns
import warnings; warnings.filterwarnings('ignore')
sns.set_theme(style='whitegrid', font_scale=1.05)
FIGDIR = 'results/figures'
import os; os.makedirs(FIGDIR, exist_ok=True)

# Fig 1: Mean trust by usage group
usage_order = ['Low Usage','Medium Usage','High Usage']
grp = df.groupby('usage_group')['Q15'].agg(['mean','sem']).reindex(usage_order)
fig, ax = plt.subplots(figsize=(6,4))
bars = ax.bar(usage_order, grp['mean'], yerr=grp['sem']*1.96,
              color=['#4393c3','#2166ac','#053061'], width=0.5,
              error_kw=dict(ecolor='#555', capsize=4, lw=1.2))
ax.axhline(3, color='#d62728', ls='--', lw=1, label='Scale midpoint (3)')
ax.set_ylim(1,5); ax.set_ylabel('Mean Trust (Q15, 1–5)')
ax.set_xlabel('Usage Group'); ax.set_title('Mean Trust by Usage Frequency Group')
for bar, val in zip(bars, grp['mean']):
    ax.text(bar.get_x()+bar.get_width()/2, bar.get_height()+0.07, f'{val:.2f}',
            ha='center', va='bottom', fontsize=10, fontweight='bold')
ax.legend(fontsize=9); fig.tight_layout()
fig.savefig(f'{FIGDIR}/fig1_trust_by_usage_group.png', dpi=150); plt.show()
```

```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import seaborn as sns
import warnings; warnings.filterwarnings('ignore')
sns.set_theme(style='whitegrid', font_scale=1.05)
FIGDIR = 'results/figures'
import os; os.makedirs(FIGDIR, exist_ok=True)

# Fig 2: Trust score distribution
counts = df['Q15'].value_counts().sort_index()
fig, ax = plt.subplots(figsize=(6,4))
ax.bar(counts.index, counts.values, color='#2166ac', edgecolor='white', width=0.7)
ax.axvline(df['Q15'].mean(), color='#d62728', ls='--', lw=1.5,
           label=f'Mean = {df["Q15"].mean():.2f}')
ax.axvline(3, color='#888', ls=':', lw=1.2, label='Scale midpoint (3)')
ax.set_xlabel('Trust Score (Q15, 1–5)'); ax.set_ylabel('Number of Students')
ax.set_title('Distribution of Trust in ChatGPT (Q15)')
ax.set_xticks([1,2,3,4,5]); ax.legend(fontsize=9); fig.tight_layout()
fig.savefig(f'{FIGDIR}/fig2_trust_distribution.png', dpi=150); plt.show()
```

```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import seaborn as sns
import warnings; warnings.filterwarnings('ignore')
sns.set_theme(style='whitegrid', font_scale=1.05)
FIGDIR = 'results/figures'
import os; os.makedirs(FIGDIR, exist_ok=True)

# Fig 3: Ethics score by language region
region_order = ['Japanese','Spanish','Italian','English','Hebrew','Arabic','Turkish']
reg = df.groupby('region')['ethics_score'].agg(['mean','sem']).reindex(region_order)
colors_reg = ['#4393c3' if m<3.3 else '#2166ac' if m<3.5 else '#053061' for m in reg['mean']]
fig, ax = plt.subplots(figsize=(7,4))
bars = ax.barh(region_order, reg['mean'], xerr=reg['sem']*1.96,
               color=colors_reg, error_kw=dict(ecolor='#555', capsize=3, lw=1))
ax.axvline(3, color='#d62728', ls='--', lw=1, label='Scale midpoint (3)')
ax.set_xlabel('Mean Ethical Concerns Score (1–5)')
ax.set_title('Ethical Concerns by Language/Region')
for bar, val in zip(bars, reg['mean']):
    ax.text(val+0.01, bar.get_y()+bar.get_height()/2, f'{val:.2f}', va='center', fontsize=9)
ax.set_xlim(1,4.2); ax.legend(fontsize=9); fig.tight_layout()
fig.savefig(f'{FIGDIR}/fig3_ethics_by_region.png', dpi=150); plt.show()
```
